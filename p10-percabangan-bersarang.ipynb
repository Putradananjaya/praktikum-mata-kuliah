Notebook ini berisi program contoh dari materi Pertemuan 10 di Lontar. Setiap bagian punya dua jenis sel:

1. **Sel formulir.** Pilih atau isi masukan di formulir sebelah kanan kode, lalu klik ▶ (atau tekan Ctrl+Enter). Ulangi dengan kombinasi lain.
2. **Sel 📋 tabel uji.** Menjalankan program untuk semua kombinasi data uji sekaligus. Tebak dulu keluarannya, baru jalankan sel ini.

**Sebelum mulai:**

- Jalankan sel **⚙️ Persiapan** di bawah satu kali.
- Bila Colab menampilkan peringatan *Notebook ini tidak ditulis oleh Google*, klik **Tetap jalankan**.
- Perubahan Anda tidak tersimpan. Untuk menyimpannya, pilih **File → Simpan salinan di Drive**.
- Kode di dalam sel 📋 memakai perulangan (Pertemuan 11) dan fungsi (Pertemuan 13). Anda belum perlu memahaminya; cukup jalankan dan baca tabelnya.

```python
#@title ⚙️ Persiapan — jalankan sel ini sekali
import io
import contextlib
import pandas as pd
from IPython.display import display


def uji_kombinasi(program, kolom, kombinasi):
    """Menjalankan setiap program untuk setiap kombinasi masukan, lalu menampilkan keluarannya."""
    baris = []
    for nilai in kombinasi:
        data = dict(zip(kolom, nilai))
        for nama, kode in program.items():
            layar = io.StringIO()
            try:
                with contextlib.redirect_stdout(layar):
                    exec(kode, dict(data))
                hasil = " / ".join(layar.getvalue().strip().splitlines()) or "(tidak ada keluaran)"
            except Exception as galat:
                hasil = type(galat).__name__
            data[nama] = hasil
        baris.append(data)
    display(pd.DataFrame(baris))


print("Siap. Sekarang Anda bisa memakai tabel uji di setiap bagian.")
```

## Bab 1 · Potongan premium dan biasa

Program contoh slide 6. Pilih status, isi total belanja, lalu klik ▶. Cobalah keempat kombinasi dan nilai tepat di ambang 1000000.

```python
#@title Program potongan premium dan biasa
status_anggota = "premium"  #@param ["premium", "biasa"]
total_belanja = 750000  #@param {type:"integer"}

if status_anggota == "premium":
    if total_belanja >= 1000000:
        persen = 20
    else:
        persen = 15
else:
    if total_belanja >= 1000000:
        persen = 15
    else:
        persen = 10

print(f"Potongan: {persen}%")
```

## Bab 2 · Letak tanda kurung

Kedua program memakai syarat yang sama, tetapi letak kurungnya berbeda. Cari kombinasi yang membuat hasilnya berbeda.

```python
#@title Kurung di depan: (premium and total) or transaksi
status = "biasa"  #@param ["premium", "biasa"]
total = 300000  #@param {type:"integer"}
jumlah_transaksi = 25  #@param {type:"integer"}

# Sulit dibaca: pembaca harus hafal tingkat operator
if status == "premium" and total >= 500000 or jumlah_transaksi >= 20:
    print("Berhak voucher (tanpa kurung)")

# Jelas: kurung menegaskan pengelompokan
if (status == "premium" and total >= 500000) or jumlah_transaksi >= 20:
    print("Berhak voucher (dengan kurung)")
```

```python
#@title Kurung di belakang: premium and (total or transaksi)
status = "biasa"  #@param ["premium", "biasa"]
total = 300000  #@param {type:"integer"}
jumlah_transaksi = 25  #@param {type:"integer"}

if status == "premium" and (total >= 500000 or jumlah_transaksi >= 20):
    print("Berhak voucher")
else:
    print("Belum berhak voucher")
```

```python
#@title 📋 Tabel uji semua kombinasi — klik ▶
program = {
    'Kurung di depan': '''
if (status == "premium" and total >= 500000) or jumlah_transaksi >= 20:
    print("Berhak voucher")
else:
    print("Belum berhak voucher")
''',
    'Kurung di belakang': '''
if status == "premium" and (total >= 500000 or jumlah_transaksi >= 20):
    print("Berhak voucher")
else:
    print("Belum berhak voucher")
''',
}
kolom = ['status', 'total', 'jumlah_transaksi']
kombinasi = [
    ('biasa', 300000, 25),
    ('biasa', 800000, 5),
    ('premium', 800000, 5),
    ('premium', 300000, 25),
    ('premium', 300000, 5),
    ('biasa', 300000, 5),
]
uji_kombinasi(program, kolom, kombinasi)
```

## Bab 3 · Kasus 2: gabungan logika dan bersarang

Kedua versi seharusnya selalu memberi hasil yang sama. Tabel di bawah membuktikannya untuk setiap kombinasi.

```python
#@title Versi gabungan logika
status = "biasa"  #@param ["premium", "biasa"]
total = 1200000  #@param {type:"integer"}

# Gabungan logika: empat kondisi berulang, rawan terlewat
if status == "premium" and total >= 1000000:
    persen = 20
elif status == "premium" and total < 1000000:
    persen = 15
elif status != "premium" and total >= 1000000:
    persen = 15
else:
    persen = 10

print(f"Potongan: {persen}%")
```

```python
#@title Versi bersarang
status = "biasa"  #@param ["premium", "biasa"]
total = 1200000  #@param {type:"integer"}

# Bersarang: struktur pemilahannya tampak jelas
if status == "premium":
    if total >= 1000000:
        persen = 20
    else:
        persen = 15
else:
    if total >= 1000000:
        persen = 15
    else:
        persen = 10

print(f"Potongan: {persen}%")
```

```python
#@title 📋 Tabel uji semua kombinasi — klik ▶
program = {
    'Gabungan': '''
# Gabungan logika: empat kondisi berulang, rawan terlewat
if status == "premium" and total >= 1000000:
    persen = 20
elif status == "premium" and total < 1000000:
    persen = 15
elif status != "premium" and total >= 1000000:
    persen = 15
else:
    persen = 10

print(f"Potongan: {persen}%")
''',
    'Bersarang': '''
# Bersarang: struktur pemilahannya tampak jelas
if status == "premium":
    if total >= 1000000:
        persen = 20
    else:
        persen = 15
else:
    if total >= 1000000:
        persen = 15
    else:
        persen = 10

print(f"Potongan: {persen}%")
''',
}
kolom = ['status', 'total']
kombinasi = [
    ('premium', 1200000),
    ('premium', 1000000),
    ('premium', 999999),
    ('biasa', 1200000),
    ('biasa', 1000000),
    ('biasa', 999999),
]
uji_kombinasi(program, kolom, kombinasi)
```

## Bab 3 · Bersarang berlapis dan penyaringan bertahap

Centang atau kosongkan ketiga syarat. Tabel di bawah menguji kedelapan kombinasi True/False sekaligus.

```python
#@title Versi bersarang berlapis
data_lengkap = True  #@param {type:"boolean"}
stok_mencukupi = False  #@param {type:"boolean"}
pembayaran_diterima = True  #@param {type:"boolean"}

# Bersarang berlapis: sulit dibaca, indentasi menumpuk
if data_lengkap:
    if stok_mencukupi:
        if pembayaran_diterima:
            print("Pesanan diproses")
        else:
            print("Pesanan dibatalkan")
    else:
        print("Stok tidak tersedia")
else:
    print("Lengkapi data terlebih dahulu")
```

```python
#@title Versi penyaringan bertahap
data_lengkap = True  #@param {type:"boolean"}
stok_mencukupi = False  #@param {type:"boolean"}
pembayaran_diterima = True  #@param {type:"boolean"}

# Penyaringan bertahap: setiap penolakan ditangani lalu selesai
if not data_lengkap:
    print("Lengkapi data terlebih dahulu")
elif not stok_mencukupi:
    print("Stok tidak tersedia")
elif not pembayaran_diterima:
    print("Pesanan dibatalkan")
else:
    print("Pesanan diproses")
```

```python
#@title 📋 Tabel uji semua kombinasi — klik ▶
program = {
    'Berlapis': '''
# Bersarang berlapis: sulit dibaca, indentasi menumpuk
if data_lengkap:
    if stok_mencukupi:
        if pembayaran_diterima:
            print("Pesanan diproses")
        else:
            print("Pesanan dibatalkan")
    else:
        print("Stok tidak tersedia")
else:
    print("Lengkapi data terlebih dahulu")
''',
    'Bertahap': '''
# Penyaringan bertahap: setiap penolakan ditangani lalu selesai
if not data_lengkap:
    print("Lengkapi data terlebih dahulu")
elif not stok_mencukupi:
    print("Stok tidak tersedia")
elif not pembayaran_diterima:
    print("Pesanan dibatalkan")
else:
    print("Pesanan diproses")
''',
}
kolom = ['data_lengkap', 'stok_mencukupi', 'pembayaran_diterima']
kombinasi = [
    (True, True, True),
    (True, True, False),
    (True, False, True),
    (True, False, False),
    (False, True, True),
    (False, True, False),
    (False, False, True),
    (False, False, False),
]
uji_kombinasi(program, kolom, kombinasi)
```
